## Section 3.2 — Discrete random variables via Uniform(0,1)

**Biased coin experiment**

We simulate a coin where  
- P(head = 1) = 0.6  
- P(tail = 0) = 0.4.  

We generate 100 samples, then 1000 samples, and look at the relative frequencies.

**Question:**  
- Why aren't the frequencies exactly 0.4 and 0.6 for n = 100?  
- What changes for n = 1000?

**Answer:**

For n = 100:
- The observed frequency of 1 (heads) will be close to 0.6, but not exactly 0.6.
- The observed frequency of 0 (tails) will be close to 0.4, but not exactly 0.4.

This is because sampling is random. Each run is just one realization of 100 independent Bernoulli(0.6) trials. There is still sampling noise. The true probabilities 0.6 and 0.4 describe the long-run behavior, not a guarantee for a single finite sample.

For n = 1000:
- The frequencies get much closer to 0.6 and 0.4.
- The random fluctuation (error) is smaller in proportion.

Why? By the Law of Large Numbers (LLN), the sample proportion
\[\hat p_n = \frac{\#\text{heads}}{n}\]
converges to the true probability \(p = 0.6\) as \(n\) grows. With n = 1000, the relative error \(|\hat p_n - 0.6|\) is typically smaller than with n = 100.

So: small samples are noisy; large samples “average out” the noise.

## Section 3.3 — Continuous distributions via inverse CDF

### Exponential(1)

We define \( X = -\ln(U) \) where \( U \sim \text{Uniform}(0,1) \). Then \(X \sim \text{Exponential}(\lambda=1)\), with mean \(E[X] = 1\) and PDF \(f(x) = e^{-x}\) for \(x \ge 0\).

When we generate 100 samples and plot a histogram:
- We expect many small values near 0.
- Large values are possible but rarer (the histogram has a long right tail).

This shape is not symmetric and not bell-shaped. It's “decaying.”

We’ll use this later when we think about skewness and the CLT.

### Normal (Gaussian)

We draw:
1. 200 samples from \(N(1,2)\) (mean 1, variance 2).
2. 100 samples each from \(N(0,1)\), \(N(0,2)\), \(N(0,3)\).

**Questions:**
- Do histograms look bell-shaped?  
- How does changing variance affect the spread?

**Answer:**

- Yes, the histograms for Normal samples are typically bell-shaped: high in the middle and tapering off symmetrically to both sides.
- The center of the bell is around the mean.
- Variance controls spread: larger variance → more spread out.

## Section 4 — Law of Large Numbers (LLN)

### 4.1 Uniform(0,1)

**Q1:** What value does the curve approach?  
**A1:** The running average approaches 0.5.

**Q2:** Compute \(E[X]\) for \(X \sim \text{Uniform}(0,1)\):

**Q3:** This is LLN in action: the sample mean converges to the expected value.

## Section 5 — Central Limit Theorem (CLT)

**Q1:** Even skewed distributions produce nearly symmetric averages for large n.  
**Q2:** The spread shrinks as \(\sigma/\sqrt{n}\).  
**Q3:** CLT explains why averages are approximately normal → allows error bars, confidence intervals, hypothesis testing.

## Summary

- LLN: averages → true mean.  
- CLT: averages → approximately Normal.  
- Monte Carlo: averages approximate integrals.
